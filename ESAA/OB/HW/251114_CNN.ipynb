# 5. 합성곱 신경망
- 딥러닝의 역전파 : 출력층->은닉층-> 입력층
- 계산과정이 복잡, 많은 자원 필요 --(해결책)-->합성곱 신경망

## 5.1.1 합성곱층의 필요성
- 합성곱층 : 이미지나 영상을 처리하는데 유용
- 이미지를 펼쳐서 분석시 데이터의 공간적 구조 무시를 방지하려고 도입됨.

## 5.1.2 합성곱 신경망 구조
합성곱+풀링층 -> 입력 이미지의 주요 특성 추출 -> 완전연결층 -> 1차원 벡터로 변환-> 소프트 함수이용 -> 최종결과 출력
1) 입력층 : 입력 이미지 데이터가 최초로 거치게 되는 계층
- 높이, 너비, 채널의 값을 갖는 3차원 데이터 <-이미지

2) 합성곱층 : 입력 데이터에서 특성을 추출하는 역할을 수행함.
- 커널이나 필터를 사용함
- 스트라이드 : 필터를 적용되는 위치 간격을 의미
- 실제 필터의 개수는 한개라는 점

3) 풀링층 : 합성곱층과 유사하게 특성맴의 차원을 다운 샘플링하여 연산량을 감소시키고, 주요한 특성 벡터를 추출하여 학습을 효과적으로 수행
- 최대 풀링
- 평균 풀링

4) 완전연결층 : 차원이 축소된 특성맵은 완전연결층에 전달되는데, 3차원 벡터에서 1차원 벡터로 펼쳐지게 됨.

5) 출력층 : 소프트맥스 활성화 함수를 사용 -> 이미지가 각 레이블에 속할 확률 값이 출력됨 -> 가장 높은 확률 값을 갖는 레이블이 최종값으로 선정됨.

## 5.1.3 1D, 2D, 3D 합성곱
- 1d 합성곱 : 필터가 시간을 축으로 좌우로만 이동할 수 있는 합성곱
- 2D 합성곱 : 필터가 방향이 2개로 움직이는 형태임.
- 3D 함성곱
- 3D 입력을 갖는 2D 합성곱 ex) LeNet-5, VGG
- 1*1 합성곱 : 채널수를 조정하여 연산량을 감소하는 효과O ex)GoogleLeNet
