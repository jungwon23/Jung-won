# 2023 전력사용량 예측 AI 경진대회
##[배경 및 주제]
https://dacon.io/competitions/official/236125/data


### 배경
전력 사용량 예측 시뮬레이션을 통한 효율적인 인공지능 알고리즘 발굴을 목표

### 주제
전력사용량 예측 AI 모델 개발

### 데이터
1. train.csv

train 데이터 : 100개 건물들의 2022년 06월 01일부터 2022년 08월 24일까지의 데이터
일시별 기온, 강수량, 풍속, 습도, 일조, 일사 정보 포함
전력사용량(kWh) 포함


2. building_info.csv

100개 건물 정보
건물 번호, 건물 유형, 연면적, 냉방 면적, 태양광 용량, ESS 저장 용량, PCS 용량


3. test.csv

test 데이터 : 100개 건물들의 2022년 08월 25일부터 2022년 08월 31일까지의 데이터
일시별 기온, 강수량, 풍속, 습도의 예보 정보


## 코드 흐름


###(1) EDA 및 전처리
- 결측치 확인 및 보간
- date_time 분리하여 hour,day,month, day_of_week feeature 생성

###(2) Feature Engineering
- 평균기온, 최대기온 변수 만들기
- outlier 행의 경우 drop
- 임시 휴무 추측 데이터 drop
- holiday : 공휴일 변수
- Sunday_holiday : 대형마트 휴무 일요일 변수
- 시간변수 푸리에 변환
- CDH:CDH 변수
- THI:THI(불쾌지수) 변수
- WCT:WCT(체감온도)변수
- 전력소비 통계량 변수<- log transformation 진행


### (3) 생성 모델
- X,Y,test 설정
- XGB 건물 유형별 단일 모델 적용
- 주요 하이퍼 파라미터 : max_depth, subsample, colsample_bytree, min_child_wieght
- KFold를 통해 7개의 fold로 데이터를 나누어 cross validaiton을 진행함
- eval metric은 SMAPE로 설정.


##[차별점 및 배운점]
### 차별점
- XGBoost 모델 사용
- 여러가지 feauture, 즉 파생변수를 만들어 성능을 높이고자함. (시간, CDH,THI,WCT 등)

### 배울점
- 여기에서 시간변수들을 단순히 month라고 하면 1,2,3,..등만 넣은 것이 아니라 시간변수 푸리에 변환이라는 것을 이용해서 주기성을 포함해 주어서 다음번에 모델링을 진행할때 이 부분도 고려해야할 것 같다.
