```python
import pandas as pd

data = pd.read_csv(file_path)
```

```python
data.columns = data.iloc[1]
data = data.drop(index=[0, 1])
data = data.reset_index(drop=True)
```

```python
list_col14 = ['D_TRA1_CASE','D_TRA2_CASE','D_TRA3_CASE',
'D_TRA4_CASE',     #0~3
'D_TRA1_1_SPOT',   # 4~9
'D_TRA1_2_SPOT',
'D_TRA1_3_SPOT',
'D_TRA1_4_SPOT',
'D_TRA1_5_SPOT',
'D_TRA1_6_SPOT',
'D_TRA2_1_SPOT',  #10~13
'D_TRA2_2_SPOT',
'D_TRA2_3_SPOT',
'D_TRA2_4_SPOT',
'D_TRA3_1_SPOT',  #14~15
'D_TRA3_2_SPOT',
'D_TRA4_1_SPOT',  #16
'D_TRA1_B7A_1',   # 17~22
'D_TRA1_B7A_2',
'D_TRA1_B7A_3',
'D_TRA1_B7B_1',
'D_TRA1_B7B_2',
'D_TRA1_B7B_3',
'D_TRA2_B7A_1',  # 23~28
'D_TRA2_B7A_2',
'D_TRA2_B7A_3',
'D_TRA2_B7B_1',
'D_TRA2_B7B_2',
'D_TRA2_B7B_3',
'D_TRA3_B7A_1',  # 29~31
'D_TRA3_B7B_1',
'D_TRA3_B7B_2',
'D_TRA4_B7A_1',  #32~33
'D_TRA4_B7B_1',]

data20 = data[list_col14]
```

```python
data20.sum()
```

```python
data20 = data20.fillna(0)
data20 = data20.astype(int)
data20 = data20.astype(object)
```

```python
data20
```

```python
spot_max = {1: 6, 2: 4, 3: 2, 4: 1}

threshold = {1: 3, 2: 3, 3: 1, 4: 1}

result = defaultdict(lambda: {"car": 0, "non_car": 0, "total": 0})

for _, row in data20.iterrows():
    for t in range(1, 5):
        case = row[f'D_TRA{t}_CASE']
        if case != 1:
            continue
        for i in range(1, spot_max[t] + 1):
            spot_col = f'D_TRA{t}_{i}_SPOT'
            region = row.get(spot_col, 0)
            if pd.isna(region) or int(region) == 0:
                continue
            if i <= threshold[t]:
                b7_col = f'D_TRA{t}_B7A_{i}'
            else:
                b7_col = f'D_TRA{t}_B7B_{i - threshold[t]}'
            mode = row.get(b7_col, 0)
            is_car = (int(mode) == 1)
            region = int(region)
            result[region]["total"] += 1
            if is_car:
                result[region]["car"] += 1
            else:
                result[region]["non_car"] += 1

summary = pd.DataFrame([
    {
        "region_code": region,
        "car_count": v["car"],
        "non_car_count": v["non_car"],
        "total_count": v["total"],
        "car_ratio": round(v["car"] / v["total"], 3)
    }
    for region, v in result.items()
]).sort_values("region_code").reset_index(drop=True)

summary
```

```python
summary.sort_values(by='car_ratio', ascending=False)
```

## 22~24 전체 자동차 의존도 계산

```python
import pandas as pd

f_all = summary[summary['total_count'] >= 4].copy()
f_2022 = region_df_2022[region_df_2022['total_count'] >= 4].copy()
f_2023 = region_df_2023[region_df_2023['total_count'] >= 4].copy()

f_all = f_all.rename(columns=lambda x: f"all_{x}" if x != "region_code" else x)
f_2022 = f_2022.rename(columns=lambda x: f"y2022_{x}" if x != "region_code" else x)
f_2023 = f_2023.rename(columns=lambda x: f"y2023_{x}" if x != "region_code" else x)

merged = pd.merge(f_all, f_2022, on="region_code", how="outer")
merged = pd.merge(merged, f_2023, on="region_code", how="outer")

merged = merged.sort_values("region_code").reset_index(drop=True)

merged
```

```python
region_df_2024_2 = summary[summary['total_count'] > 5].reset_index(drop=True)
region_df_2022_2 = region_df_2022[region_df_2022['total_count'] > 5].reset_index(drop=True)
region_df_2023_2 = region_df_2023[region_df_2023['total_count'] > 5].reset_index(drop=True)
```

```python
top20_all = region_df_2024_2.sort_values("car_ratio", ascending=False).head(20).reset_index(drop=True)

top20_2022 = region_df_2022_2.sort_values("car_ratio", ascending=False).head(20).reset_index(drop=True)

top60_2023 = region_df_2023_2.sort_values("car_ratio", ascending=False).head(60).reset_index(drop=True)

top20_all
top20_2022
top60_2023
```

```python
import pandas as pd

df_all = pd.concat([
    region_df_2024_2,
    region_df_2022_2,
    region_df_2023_2
], ignore_index=True)

merged = df_all.groupby('region_code', as_index=False).agg({
    'car_count': 'sum',
    'non_car_count': 'sum',
    'total_count': 'sum'
})
merged['car_ratio'] = merged['car_count'] / merged['total_count']
merged = merged.sort_values('region_code').reset_index(drop=True)

merged
```
