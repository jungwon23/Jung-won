## 9.2 가우시안 혼합
* 가우시안 혼합 모델: 샘플이 파라미터가 알려지지 않은 여러 개의 혼합된 가우시안 분포에서 생성되었다고 가정하는 확률 모델
* GaussianMixture 클래스에 구현됨.
* 기댓값-최적화 알고리즘을 사용함

(1) 기댓값 단계: 샘플을 클러스터에 할당

(2) 최대화 단계: 클러스터를 업데이트

(3) 소프트 클러스터 할당을 사용함

(4) 책임: 클러스터에 속할 추정 확률

### 9.2.1 가우시안 혼합을 사용한 이상치 탐지

* 이상치 탐지는 보통과 많이 다른 샘플을 감지하는 작업

* 사용할 밀도 임곗값을 정해야 한다.

* 특이치 탐지는 이상치 탐지와 달리 이상치로 오염되지 않은 깨끗한 데이터셋에서 훈련한다.

### 9.2.2 클러스터 개수 선택하기

* BIC, AIC

*확률: 파라미터 theta인 확률 모델이 주어지면, 미래 출력 x가 얼마나 그럴듯한지 설명한다.

* 가능도: 출력 x를 알고 있을 때, 특정 파라미터 값 theta가 얼마나 그럴듯한지 설명한다.

### 9.2.3 베이즈 가우시안 혼합 모델

*BayesGaussianMixture 클래스 사용

*클러스터 개수는 n_components를 최적의 클러스터 개수보다 크다고 믿을 만한 값으로 지정한다.

* 자동으로 불필요한 클러스터를 제거

### 9.2.4 이상치 탐지와 특이치 탐지를 위한 다른 알고리즘

* 이상치 탐지와 특이치 탐지 전용으로 사용할 수 있는  다른 알고리즘

* PCA :보통 샘플의 재구성 오차와 이상치의 재구성 오차를 비교하면 일반적으로 후자가 훨씬 크다.

* Fast-MCD :EllipticEnvelope 클래스에서 구현된 이 알고리즘은 이상치 감지에 유용

* 아이솔레이션 포레스트 :특히 고차원 데이터셋에서 이상치 감지를 위한 효율적인 알고리즘

* LOF :이상치 탐지에 좋음

* on-class SVM :특이치 탐지에 잘 맞는다.

#[코드필사]

```python
import numpy as np
from sklearn.datasets import make_moons

X, y = make_moons(n_samples=1000,noise=0.05)
```

```python
from sklearn.mixture import GaussianMixture

gm = GaussianMixture(n_components=3, n_init=10)
gm.fit(X)
```

```python
gm.means_
```

```python
gm.covariances_
```

```python
gm.converged_
```

```python
gm.n_iter_
```

```python
gm.predict(X)
```

```python
gm.predict_proba(X)
```

```python
X_new, y_new = gm.sample(6)
X_new
```

```python
y_new
```

```python
gm.score_samples(X)
```

```python
densities = gm.score_samples(X)
density_threshold = np.percentile(densities,4)
anomalies = X[densities < density_threshold]
```

```python
gm.bic(X)
```

```python
gm.aic(X)
```

```python
from sklearn.mixture import BayesianGaussianMixture
bgm = BayesianGaussianMixture(n_components=10, n_init=10)
bgm.fit(X)
np.round(bgm.weights_,2)
```
