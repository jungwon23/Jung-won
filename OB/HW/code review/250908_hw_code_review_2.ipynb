# 채무 불이행 여부 예측 해커톤: 불이행의 징후를 찾아라!
**목표**

다양한 금융 데이터를 활용하여 채무 불이행 가능성을 예측하는 AI 알고리즘을 개발하는 것

**주제**

채무 불이행 여부 예측 AI 알고리즘 개발

**데이터**
* train.csv: 샘플별 고유 ID (10,000 rows), 채무와 관련된 개인 신용 데이터, 채무 불이행 여부
* train.csv: 샘플별 고유 ID (10,000 rows), 채무와 관련된 개인 신용 데이터

## [코드 리뷰]
**(1)EDA**
- 범주형 변수 : boxtplot
- 수치형 변수: heatmap, boxtplot

**(2) Feature Engineering**
- 대출 상환기간(장기/단기)-> 더미 데이터 생성
- 대출목적 -> 생활비, 주거, 부채통합, 기타 4가지 카테고리로 축소
-주거형태 -> 주택담보 대출 카테고리 통합
- 파생변수 생성: 자금문제발생횟수, 위험수준, 미상환 금액 평균, 연간 개설 계좌 수

-> 이후 기존 변수 삭제 진행

- 로그변환 : 금액 변수
- 이상치 :IQR 계산 -> 이상치를 상한과 하한으로 각각 대체함
- 정규화 : 수치형 변수 StandardScaler 적용
- 다중공선성 확인

**(3)모델링**
- 로지스틱 회귀 모델 생성
- GridSearchCV를 이용해서 최적의 하이퍼 파라미터 튜닝을 진행함
- 모델 생성: logistic 회귀 모형에 penalty='l2'부여
- 모델 검증 : accuracy_score, ROC-AUC 점수 확인
- 변수별 로지스틱 회귀를 이용한 가중치 부여

# [차별점 및 배울점]
## [차별점]
- 이상치를 IQR의 상한과 하한을 이용해서 대체한 점
- 로지스틱 회귀분석을 진행한 후 여러 평가 지표를 이용해서 평가를 진행한 점

## [배울점]
- 모델의 평가 지표를 accuracy score 외에도 여러 평가 지표를 이용해서 확인할 필요가 있다는 점
- 기존의 변수들도 유사한 카테고리끼리 묶어 축소를 하여서 분석을 진행할 수 있다는 부분을 알게됨
