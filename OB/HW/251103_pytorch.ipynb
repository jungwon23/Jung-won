##1.1 인공지능, 머신러닝과 딥러닝
- 인공지능 : 인간의 지능을 모방하여 사람이 하는일을 컴퓨터가 할 수 있도록 하는 기술
- 인공지능 : 머신러닝+딥러닝
1) 머신러닝 : 주어진 데이터 인간 처리<- 데이터 특징을 스스로 추출X
2) 딥러닝: 인간이 하던 작업 생략, 컴퓨터가 스스로 분석한 후에 답을 도출

##1.2 머신러닝이란
### 1.2.1 머신러닝 학습 과정
-과정: 학습단계 -> 예측단계
-머신러닝 구성요소 : 데이터 + 모델

### 1.2.2 머신러닝 학습 알고리즘
1) 지도학습 : 정답을 알려주고 컴퓨터가 학습
2) 비지도학습: 정답을 알려주짙, 범주화하여 예측
3) 강화학습: 보상이 커지는 행동을 자주, 줄어드는 행동을 덜하도록 학습

##1.3 딥러닝이란
-딥러닝: 심층 신경망 이론을 기반으로 고안된 머신러닝 방법의 일종

### 1.3.1 딥러닝 학습 과정
-데이터 준비 -> 모델 정의(신경망 생성) -> 모델 컴파일(활성화 함수, 손실함수, 옵티마이저 선택) -> 모델 훈련(배치,에포크 선택) -> 모델 예측
- 핵심구성요소 : 신경망과 역전파 <- 심층 신경망을 사용

### 1.3.2 딥러닝 학습 알고리즘
1) 지도학습- 합성곱 신경망(이미지분류),순환신경망(시계열분류)
2) 비지도학습- 워드임베딩, 군집
3) 전이학습
4) 사전 학습 모델
5) 강화학습

- 파이토치 API -> 파이토치 엔진(다차원 텐서와 자동미분을 처리) -> 텐서에 대한 연산 처리
1) 파이토치 API : 사용자 인터페이스 제공, 실제계산 X

## 2.2 파이토치 기초 문법
### 2.2.1 텐서 다루기
-텐서 생성 및 변환: torch.numpy()
-텐서의 인덱스 조작
-텐서의 연산 및 차원 조작 : 텐서간의 타입이 다르면 연산이 불가능, 텐서 조작 view, stack,cat, t, transpose이용

### 2.2.2 데이터 준비
- 파이썬 라이브러리를 이용, 파이토치에서 제공하는 데이터를 이용하는 방법
- 커스텀 데이터셋: 데이터를 한번에 다 부르지 않고 조금씩 나누어 불러서 사용하는 방식
- 토치비전: 파이포치에서 제공하는 데이터셋이 모여있는 패키지
- 데이터셋 다운: request 라이브러리 설치

### 2.2.3모델 정의
- 계층, 모듈, 모델
-nn.Module()을 상속하여 정의 하는 방법 : init(), forward()함수 이용
-nn.Sequential

### 2.2.4 모델의 파라미터 정의
- 손실함수 : BCELoss, CrossEntropy, MSELoss
- 옵티마이저
- 학습률 스케줄러: 미리 지정한 횟수의 에포크를 지날때마다 학습률을 감소시켜줌.
-지표: 훈련과 테스트 단계 모니터링

### 2.2.5 모델 훈련
1) optimizer.zero_grad() 메서지 : 기울기 초기화
2) backward() : 새로운 기울기 값이 이전 기울기 값에 누적하여 계산

### 2.2.6 모델 평가
1) 함수를 이용하여 평가
2) 모듈을 이용하여 평가

### 2.2.7 훈련과정 모니터링
- model.train() : 훈련 데이터셋에서 사용하며 모델 훈련이 진행됨.
- model.eval(): 모델 평가시 모든 노드 사용

```python
import torch.nn as nn
class MLP(nn.Module):
  def __init__(self):
    super(MLP, self).__init__()
    self.layer1=nn.Sequential(
        nn.Conv2d(in_channels=3, out_channels=64, kernel_size=5),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(2))

    self.layer2=nn.Sequential(
        nn.Conv2d(in_channels=64, out_channels=30, kernel_size=5),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(2))

    self.layer3=nn.Sequential(
        nn.Linear(in_features=30*5*5, out_features=10, bias=True),
        nn.ReLU(inplace=True))

    def forward(self, x):
      x=self.layer1(x)
      x=self.layer2(x)
      x=x.view(x.shape[0], -1)
      x=self.layer3(x)
      return x

model=MLP() # 모델에 대한 객체 생성

print('Printing childern\n--------')
print(list(model.children()))
print('\n\nPrinting Modules\n-------')
print(list(model.modules()))
```
