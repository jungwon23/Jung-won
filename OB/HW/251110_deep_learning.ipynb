## 4.2 딥러닝 구조
- 딥러닝: 여러 층을 가진 인공 신경망을 사용하여 학습을 수행하는 것

### 4.2.1 딥러닝 용어
- 딥러닝은 입력층, 출력층과 2개 이상의 은닉층으로 구성
- 입력층, 은닉층, 출력층
- 가중치: 입력 값이 연산 결과에 미치는 영향력을 조절하는 요소
- 활성화 함수 : 시그모이드, 하이퍼볼릭 탄젠트, 렐루 함수 등이 존재
- 손실 함수:평균 제곱 오차,크로스 엔트로피 오차

### 4.2.2 딥러닝 학습
- 딥러닝 학습은 크게 순전파와 역전파 2 단계로 진행
- 순전파: 네트워크에 훈련 데이터가 들어올 때 발생
- 역전파: 손실 함수 비용이 0에 가깝도록 하기 위해 모델이 훈련을 반복하면서 가중치를 조정해나가는데, 이 과정에서 손실이 계산되면 정보는 역으로 출력층에서 입력층으로 전달되는 현상이 나타난다.

### 4.2.3 딥러닝의 문제점과 해결 방안

- 과적합 문제 -(해결법)-> 드롭 아웃
- 기울기 소멸 문제 -(해결법)-> 렐루 활성화 함수
- 성능 저하 문제 -(해결법) -> 확률적 경사 하강법, 미니 배치 경사 하강법

### 4.2.4 딥러닝을 사용할 때 이점
1) 특성 추출 : 데이터 별로 특징을 찾고 이를 바탕으로 데이터를 벡터로 변환하는 작업
2) 빅데이터의 효율적 활용

##4.3 딥러닝 알고리즘
### 4.3.1 심층 신경망
-심층신경망: 입력층과 출력층 사이에 다수의 은닉층을 포함하는 인공 신경망

### 4.3.2 합성곱 신경망
- 합성곱층과 풀링층을 포함하는 이미지 처리 성능이 좋은 신경망 알고리즘
- 대표적으로 LENEt-5, AlexNet이 있음.

### 4.3.3 순환 신경망
- 시계열 데이터 같은 시간 흐름에 따라 변화하는 데이터를 학습하기 위한 인공신경망
- 특징: 시간성을 가진 데이터가 많고, 시간성 정보를 잘 다룸, 데이가 동적이고 길이가 가변적임, 매우 긴 데이터를 처리하는 연구에 활발히 진행
- 기울기 소멸 문제 -(해결방법) -> LSTM

### 4.3.4 제한된 볼츠만 머신
- 가시층과 은닉층으로 구성된 모델
- 가시층은 은닉층과만 연결이 된다.

### 4.3.5 심층 신뢰 신경망
- 입력과 은닉층으로 구성된 제한된 볼츠만 머신을 블록 처럼 여러층으로 쌓은 형태로 연결된 신경망
- 제한된 볼츠만 머신 사전 훈련 -> 첫번째 층 고정, 두번째 층 사전 훈련 -> 원하는 만큼 쌓아올려 완성




```python
import torch

class Net(torch.nn.Module):
    def __init__(self, n_feature, n_hidden, n_output):
        super(Net, self). __init__()
        self.hidden = torch.nn.Linear( n_feature, n_hidden)
        self.relu = torch.nn.ReLu(inplace=True )
        self.out = torch.nn.Linear( n_hidden , n_output)
        self.softmax = torch.nn.Softmax(dim=n_output)
    def forward (self, x):
        x = self.hidden(x)
        x = self.relu(x)

        x = self.out(x)
        x = self.softmax(x)
        return x
```
